# 01 — Data Preparation

Loads and merges the six IBM Telco Customer Churn files, removes leak
columns, encodes categoricals, and scales numerics. Produces the final
feature matrix (`X`), target (`y`), and segment dataframe used by
notebooks 02–05.

**Maps to paper Section 9.1.**

## Inputs
Place the six `.xlsx` files in `data/raw/`. See `data/README.md` for
download instructions.

## Outputs
- `outputs/tables/X_features.csv`
- `outputs/tables/y_target.csv`
- `outputs/tables/segments.csv`


## 1. Setup

```python
# Make the project's src/ importable. If you're in Colab, clone the repo first.
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore")
print("Libraries loaded.")
```

## 2. Load the six Excel files

```python
from src.data_loader import load_telco_files

DATA_DIR = "../data/raw"   # adjust if running in Colab
frames = load_telco_files(DATA_DIR)
```

Peek at each file to confirm everything loaded:

```python
for key, df in frames.items():
    print(f"\n{key}: {df.shape}")
    print(df.columns.tolist()[:8], "...")
```

## 3. Merge into one dataframe

```python
from src.data_loader import merge_telco_data

df = merge_telco_data(frames)
print(f"\nFinal merged shape: {df.shape}")
df.head()
```

## 4. Prepare features

```python
from src.preprocessing import prepare

X, y, segments_df = prepare(df, drop_leaks=True)
print(f"X: {X.shape}")
print(f"y: {y.shape}")
print(f"segments_df: {segments_df.shape}")
```

## 5. Verify the segment columns

```python
print("Segment columns:")
print(segments_df.head())
print("\nDistributions:")
for col in segments_df.columns:
    print(f"\n{col}:")
    print(segments_df[col].value_counts())
```

## 6. Quick sanity-check on the target

```python
print(f"Total customers: {len(y)}")
print(f"Churn rate: {y.mean():.2%}")
print(f"\nClass counts:\n{y.value_counts()}")
```

## 7. Save processed data

```python
OUT_DIR = Path("../outputs/tables")
OUT_DIR.mkdir(exist_ok=True, parents=True)

X.to_csv(OUT_DIR / "X_features.csv", index=False)
y.to_csv(OUT_DIR / "y_target.csv", index=False)
segments_df.to_csv(OUT_DIR / "segments.csv", index=False)
print(f"Saved to {OUT_DIR.resolve()}")
```
