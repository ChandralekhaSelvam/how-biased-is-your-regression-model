## 2. Load the analysis dataframe from notebook 04

## 3. Per-segment metrics for Contract

## 4. Figure 3 — Segment-Level Accuracy and Mean Error

## 5. Figure 4 — Accuracy by Gender

```python
from src.visualizations import plot_accuracy_by_gender

gender_metrics = segment_metrics_table(df_analysis, "Gender")
print(gender_metrics.round(4))

fig = plot_accuracy_by_gender(gender_metrics)
savefig(fig, "../outputs/figures/fig4_accuracy_by_gender.png")
fig.show()
```

## 6. Figure 5 — ROC curves by Contract type

```python
from src.visualizations import plot_roc_by_contract

fig = plot_roc_by_contract(df_analysis,
                           proba_col="fair_pred_proba",
                           title="ROC Curves by Contract - Fair Model (λ=2)")
savefig(fig, "../outputs/figures/fig5_roc_by_contract.png")
fig.show()
```

## 7. Compute fairness metrics across segments

```python
from src.metrics import (
    equalized_odds_difference,
    demographic_parity_difference,
)

dp = demographic_parity_difference(
    df_analysis["fair_pred_class"], df_analysis["Contract"]
)
eo = equalized_odds_difference(
    df_analysis["y_true"], df_analysis["fair_pred_class"], df_analysis["Contract"]
)

print(f"Demographic Parity Difference (Contract): {dp:.4f}")
print(f"Equalized Odds Difference   (Contract): {eo:.4f}")
```
